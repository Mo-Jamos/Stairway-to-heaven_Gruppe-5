# Stairway to Heaven – YOLO in Google Colab trainieren

Kostenlose Grafikkarte von Google, Training dauert ca. 10–30 Minuten statt Stunden.

**Vorher auf dem Laptop:** `python tools/yolo_train.py --zip` erzeugt `data/yolo/stairs_dataset.zip`.

**Hier:** Menü *Laufzeit → Laufzeittyp ändern → T4-GPU* wählen, dann die Zellen der Reihe nach mit ▶ ausführen.

```
# 1. YOLO installieren und Grafikkarte prüfen
!pip install -q ultralytics
import torch; print("Grafikkarte:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "KEINE - Laufzeittyp auf GPU stellen!")
```

```
# 2. stairs_dataset.zip hochladen (Dateiauswahl öffnet sich) und entpacken
from google.colab import files
up = files.upload()
!rm -rf /content/dataset && unzip -q stairs_dataset.zip -d /content && ls /content/dataset
```

```
# 3. Trainieren (Epochen bei Bedarf erhöhen, z. B. 150)
from ultralytics import YOLO
model = YOLO("yolo11n.pt")
model.train(data="/content/dataset/data_colab.yaml", epochs=100, imgsz=1024, batch=16,
            patience=30, fliplr=0.5, flipud=0.5, degrees=0.0, hsv_h=0.0, hsv_s=0.0, hsv_v=0.2,
            project="/content/runs", name="stairs")
```

```
# 4. Ergebnis ansehen (Lernkurven)
from IPython.display import Image, display
display(Image(filename=str(model.trainer.save_dir / "results.png")))
```

```
# 5. bestes Modell herunterladen
#    Auf dem Laptop speichern als:  stairway_prototype/models/stairs_yolo.pt
import shutil
shutil.copyfile(model.trainer.best, "/content/stairs_yolo.pt")
files.download("/content/stairs_yolo.pt")
```
